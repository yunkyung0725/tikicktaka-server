````python
def preprocessing(document): #특수문자 기본 전처리 함수(토큰화할때 마침표제거, 숫자변환)
    docu = []
    for doc in document:
        doc = re.sub('[^가-힣0-9a-zA-Z\s.]','',doc) #한글·숫자·알파벳을 제외하고 모두 제거
        doc = re.sub('[-=+,.#/\?:^$@*\"※~&%ㆍ!』\\‘|\(\)\[\]\<\>`\'…》]','',doc) #특수문자 제거
        doc = re.sub('[0-9]+','num',doc) #숫자를 special token num으로 치환
        doc = doc.lower() #소문자로 통일
        pattern = re.compile(r'\s+') #중복띄어쓰기 제거
        doc = re.sub(pattern,' ',doc) #중복띄어쓰기 제거
        doc = doc.replace('\xa0','')
        doc = doc.replace('\n','')
        docu.append(doc)

    return docu
````

````python
def mecab_morphs(text): # 형태소 단위로 tokenization
    morphs = []
    pattern = re.compile(".*\t[A-Z]+")
    temp = [tuple(pattern.match(token).group(0).split("\t")) for token in mecab.parse(text).splitlines()[:-1]]
    for token in temp:
        morphs.append(token[0])
    return morphs

def mecab_except_josa(text): #조사와 어미 tag 제외하고 반환
    morphs = []
    pattern = re.compile(".*\t[A-Z]+")
    temp = [tuple(pattern.match(token).group(0).split("\t")) for token in mecab.parse(text).splitlines()[:-1]]
    josa_ls = ['EP','EF','EC','JKS','JKC','JKG','JKO','JKB','JC','JX']
    for token in temp:
        if token[1] in josa_ls:
            pass
        else:
            morphs.append(token[0])
    return morphs

def mecab_nouns(text): #명사만 반환
    morphs = []
    pattern = re.compile(".*\t[A-Z]+")
    temp = [tuple(pattern.match(token).group(0).split("\t")) for token in mecab.parse(text).splitlines()[:-1]]
    for token in temp:
        if token[1] == 'NNG':
            morphs.append(token[0])
    return morphs
````

````python
morphs = [mecab_morphs(x) for x in text_ls]
nouns = [mecab_nouns(x) for x in text_ls]
ex_josa = [mecab_except_josa(x) for x in text_ls]

morphs = [element for array in morphs for element in array]
nouns = [element for array in nouns for element in array]
ex_josa = [element for array in ex_josa for element in array]
````

````python
# prompt: # 불용어 처리
# stopPos = ['Suffix','Punctuation','Josa','Foreign','Alpha','Number']
# #최빈어 조회: 최빈어를 조회하여 불용어 제거 대상을 선정
# from collections import Counter
# Counter(oktTag).most_common()여기 코드에서 위의 okt_tokens되어있는 거를 개수전체 다세고 품사를 설정 및 지정을 해서 쭉 뽑아서 알려줘.최빈어 출력도 해주고 전체 단어들의 품사르르 태깅해서 갯수도 세어주고 쭉 뽑아서 하나의 csv파일로 줘도 돼

import pandas as pd
from konlpy.tag import Okt
from collections import Counter
import csv

try:
    df = pd.read_csv('/content/reviews_all.csv.csv')

    okt = Okt()
    if 'content' in df.columns:
        df['okt_tokens'] = df['content'].apply(lambda x: okt.morphs(str(x)))
        df['oktTag'] = df['okt_tokens'].apply(lambda x: okt.pos(' '.join(x)))

        # 모든 토큰과 품사 태그를 하나의 리스트에 저장
        all_tokens_with_tags = []
        for tokens in df['oktTag']:
            all_tokens_with_tags.extend(tokens)

        # 전체 토큰의 품사 태그별 빈도수 계산
        tag_counts = Counter(tag for _, tag in all_tokens_with_tags)
        print("전체 단어의 품사 태깅 결과:")
        print(tag_counts)

        # 최빈어 출력 (상위 20개)
        top_20_tokens = Counter(token for token, tag in all_tokens_with_tags).most_common(20)
        print("\n최빈어 20개 (빈도수 순):")
        for token, count in top_20_tokens:
            print(f"{token}: {count}")

        # CSV 파일로 저장
        with open('token_pos_counts.csv', 'w', newline='', encoding='utf-8') as csvfile:
            writer = csv.writer(csvfile)
            writer.writerow(['Token', 'POS', 'Count'])  # 헤더 추가
            for (token, pos), count in Counter(all_tokens_with_tags).items():
                writer.writerow([token, pos, count])

    else:
        print("Error: 'content' column not found in the DataFrame.")

except FileNotFoundError:
    print("Error: File not found. Please check the file path.")
except pd.errors.ParserError:
    print("Error: Unable to parse the file. Please check the file format.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
````

````python
# prompt:

import pandas as pd
from konlpy.tag import Okt
import os

# Check if the file exists before attempting to read it
if not os.path.exists('/content/reviews_all.csv.csv'):
    print("Error: reviews_all.csv.csv not found in /content/.")
else:
    try:
        df = pd.read_csv('/content/reviews_all.csv.csv')
        print(df.head())  # Print the first few rows to verify

        okt = Okt()

        def preprocess_text(text):
            if isinstance(text, str):  # Check if the input is a string
                # 1. Tokenization with POS tagging
                tokens = okt.pos(text, stem=True)  # stemming is applied for better accuracy

                # 2. Remove unnecessary POS tags
                cleaned_tokens = [word for word, tag in tokens if tag not in ['Josa', 'Eomi', 'Punctuation', 'Foreign']]

                # 3. Join the cleaned tokens back to a string
                return " ".join(cleaned_tokens)
            else:
                return ""  # Handle cases where text is not a string (e.g., NaN)


        df['cleaned_content'] = df['content'].apply(preprocess_text)
        print(df[['content', 'cleaned_content']].head())

    except pd.errors.ParserError:
        print("Error: Unable to parse the file. Please check the file format.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
````

````python
from tokenizers import Tokenizer, models, trainers, pre_tokenizers

# BPE 토크나이저 생성
tokenizer = Tokenizer(models.BPE())

# 트레이너 설정 (BPE 학습)
trainer = trainers.BpeTrainer(vocab_size=5000, min_frequency=2)

# 학습 데이터
data = ["은행어플은 사용하기 어려워요", "신한은행 어플은 빠르다"]

# 전처리기 설정 (띄어쓰기 단위로 나누기)
tokenizer.pre_tokenizer = pre_tokenizers.Whitespace()

# BPE 학습 실행
tokenizer.train_from_iterator(data, trainer)

# 테스트
output = tokenizer.encode("기업은행 어플이 느려요")
print(output.tokens)  # ['기업', '은행', '어플', '이', '느려', '요']
````

````python
# prompt: labeled_output.csv 이제 이파일에서 BERT모델을 적용하여 감성분석을 수행

!pip install transformers

import pandas as pd
import torch
from transformers import BertTokenizer, BertForSequenceClassification
from google.colab import files
import ast

# Upload the 'encoded_file.csv' file
uploaded = files.upload()

for fn in uploaded.keys():
  print('User uploaded file "{name}" with length {length} bytes'.format(
      name=fn, length=len(uploaded[fn])))

# Read the CSV file into a pandas DataFrame

try:
    df_encoded = pd.read_csv(fn, encoding='utf-8-sig')
except pd.errors.ParserError:
    try:
      df_encoded = pd.read_csv(fn, encoding='cp949')
    except:
      print("Error: Could not parse the CSV file. Check the file format and encoding.")

# Convert the 'integer_encoded' column back to lists
# 'integer_encoded' 열을 리스트로 변환
df_encoded['integer_encoded'] = df_encoded['integer_encoded'].apply(ast.literal_eval)


# Load pre-trained KoBERT tokenizer and model
#'integer_encoded' 열의 문자열을 실제 리스트 객체로 변환합니다.
model_name = "monologg/kobert"
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertForSequenceClassification.from_pretrained(model_name, num_labels=2) # 2 labels for binary classification


# Function to perform sentiment analysis
#사전 학습된 KoBERT 토크나이저와 이진 분류를 위한 모델을 로드합니다.
def analyze_sentiment(text):
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True)
    outputs = model(**inputs)
    logits = outputs.logits
    predicted_class = torch.argmax(logits, dim=1).item()
    return predicted_class # 0 for negative, 1 for positive


# Apply sentiment analysis to each review
#주어진 텍스트에 대해 감정 분석을 수행하는 함수를 정의하고 각 리뷰에 감정분석을 적용한다.
df_encoded["sentiment"] = df_encoded["content"].apply(analyze_sentiment)

# Save the results to a new CSV file
df_encoded.to_csv("sentiment_results.csv", encoding="utf-8-sig", index=False)

# Download the results file
files.download("sentiment_results.csv")
print(df_encoded.head())
````
